```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.util import ngrams
from collections import Counter
import re
```

```python
# Download necessary NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python

# Initialize lemmatizer and stop words
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Load dataset
combined_df = pd.read_csv(r"C:/BDA proj/cleaned_datasets/combined_data.csv")

# Preprocessing function to clean text
# Preprocessing function to clean text
def preprocess_text(text):
    if pd.isnull(text) or not isinstance(text, str):  # Handle NaN and non-string values
        return ""
    
    # Replace slashes, commas, and special characters with spaces
    text = re.sub(r'[\\/|,]', ' ', text)
    
    # Handle improperly concatenated words (e.g., CamelCase, mixed letters and numbers)
    text = re.sub(r'(?<=[a-z])(?=[A-Z])', ' ', text)
    text = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', text)
    text = re.sub(r'([a-zA-Z])(\d)', r'\1 \2', text)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)  # Remove non-alphabetic characters
    
    # Tokenize, lowercase, and lemmatize
    tokens = word_tokenize(text.lower())
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word.isalpha() and word not in stop_words]

    # Join tokens back into a single string
    return ' '.join(tokens)

# Apply preprocessing
combined_df['cleaned_skills'] = combined_df['skills'].apply(preprocess_text)
combined_df['processed_location'] = combined_df['locations'].apply(preprocess_text)
combined_df['processed_industry'] = combined_df['industry'].apply(preprocess_text)
combined_df['processed_roles'] = combined_df['roles'].apply(preprocess_text)


# Display first 5 cleaned rows
print(combined_df['cleaned_skills'].head())
```

```python
# Function to extract n-grams
def extract_ngrams(text, ngram_size=2):
    if not isinstance(text, str) or text.strip() == "":  # Ensure valid string
        return []
    tokens = word_tokenize(text)
    n_grams = ngrams(tokens, ngram_size)
    return [' '.join(gram) for gram in n_grams]

# Extract 2-grams (bigrams) and 3-grams (trigrams) from the cleaned dataset
all_bigrams = []
all_trigrams = []

for text in combined_df['cleaned_skills']:
    bigrams = extract_ngrams(text, ngram_size=2)
    trigrams = extract_ngrams(text, ngram_size=3)
    all_bigrams.extend(bigrams)
    all_trigrams.extend(trigrams)

# Count frequency of bigrams and trigrams
bigram_counts = Counter(all_bigrams)
trigram_counts = Counter(all_trigrams)

# Display the most common bigrams and trigrams
print("Most common bigrams:")
print(bigram_counts.most_common(20))

print("Most common trigrams:")
print(trigram_counts.most_common(20))
```

```python
# Combine bigrams and trigrams to form the skill dictionary
skills_dict = set(bigram_counts.keys()) | set(trigram_counts.keys())

# Display the generated skill dictionary (first 20 skills)
print("Generated Skill Dictionary:")
print(list(skills_dict)[:20])
```

```python
# Function to match skills from the dictionary in the text
def match_skills(text, skills_dict):
    if pd.isnull(text):  # Handle NaN values
        return ''
    
    # Convert text to string if it's not already
    text = str(text)
    
    matched_skills = []
    for skill in skills_dict:
        if skill in text:  # Check if the skill is present in the text
            matched_skills.append(skill)
    return ', '.join(matched_skills)

# Apply skill matching and add as a new column
combined_df['extracted_skills'] = combined_df['cleaned_skills'].apply(lambda x: match_skills(x, skills_dict))

# Save the dataset with all original and new columns in the current directory
output_filename = "processed_combined_data.csv"
combined_df.to_csv(output_filename, index=False)

print(f"Processed dataset saved as {output_filename}")
```
